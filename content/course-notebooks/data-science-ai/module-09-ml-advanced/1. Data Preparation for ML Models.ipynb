##**Handling missing values: Imputation techniques**

missing-data-pattern-8597340-6842610.gif

Sometimes, our data has missing values! 😱 But don't worry, we can handle them using imputation techniques. Here are some strategies:


### **Why missing value imputation?**

Missing value imputation is important because many data analysis techniques and machine learning models require a complete dataset to work properly. Here's a simple explanation:

Imagine you're trying to complete a jigsaw puzzle, but some pieces are missing. Without all the pieces, the picture isn't complete, and you can't fully understand what it's supposed to show. In the same way, missing values in data can lead to incomplete information, which makes it hard to analyze the data accurately. Here's why filling in these missing values (or imputation) is important:

1. **Avoiding Errors:** Many computer programs that analyze data can't handle missing values and will give errors if they're there. By filling these gaps, we can use these programs without problems.

2. **Better Decisions:** When we fill in missing values correctly, we get a more complete picture of the data. This helps us make better decisions or predictions using the data.

3. **Fair Comparisons:** If some data is missing for one part of your analysis but not another, it can lead to unfair or incorrect comparisons. By filling in these gaps, you make sure that you're comparing things fairly.

4. **Understanding Relationships:** To understand how different parts of your data are related to each other, you need complete data. Missing values can hide these relationships.

5. **Improving Accuracy:** In machine learning, models learn from data to make predictions. If there are missing values, the model might not learn well and could make less accurate predictions.

However, it's important to fill in these gaps carefully. If we guess the missing values wrong, it can lead to incorrect conclusions. It's like guessing the wrong piece in our puzzle – it can make the whole picture look different from what it should be.

###1️⃣ **Mean Imputation:**

✏️️Replace missing values with the mean of the available data. It's like filling in the gaps with the average value! We use SimpleImputer with the strategy set to 'mean' to replace missing values with the mean value of the respective column.

**What does `SimpleImputer` do?**


The `SimpleImputer` class in scikit-learn (often referred to as sklearn, which is its Python package name) is a tool used for handling missing values in datasets, which is a common issue in real-world data.

Here are the key features and functionalities of `SimpleImputer`:

1. **Missing Values Handling:** `SimpleImputer` replaces missing values, which are represented as NaNs (Not a Number) or other placeholders, in a dataset.

2. **Imputation Strategies:** It offers several strategies for imputing missing values, including:
   - `mean`: Replaces missing values using the mean along each column. Can only be used with numeric data.

```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer

# Create the DataFrame
data = {
    'Name': ['John', 'Amy', 'Chris', 'Mike', 'Sarah', 'Tom'],
    'Age': [25, 32, None, 40, 28, None],
    'Salary': [50000, 60000, None, None, 45000, 70000],
    'Email': ['john@gmail.com', 'amy@gmail.com', 'chris@', None, None, '']
}

df = pd.DataFrame(data)

# Display the DataFrame with missing values
print("DataFrame with Missing Values:")
print(df)

# Imputation using Mean for numeric columns
df_mean=df.copy()
numeric_columns = ['Age', 'Salary']
imputer_mean = SimpleImputer(strategy='mean')
df_mean[numeric_columns] = imputer_mean.fit_transform(df_mean[numeric_columns])

# Display the imputed DataFrame
print("\nImputed DataFrame using Mean:")
print(df_mean)
```

   - `median`: Replaces missing values using the median along each column. This is a better choice than mean for skewed numeric distributions and can only be used with numeric data.


In summary, `SimpleImputer` is a practical and flexible tool for dealing with missing values in machine learning datasets, making it easier to prepare data for model training and prediction.

#### 2️⃣ **Median Imputation:** 📊

Similar to mean imputation, but we use the median instead. This robust method can be useful when outliers are present. We use SimpleImputer with the strategy set to 'median' to replace missing values with the median value of the respective column.

```python
print("DataFrame with Missing Values:")
print(df)
# Imputation using Median
df_median=df.copy()
numeric_columns = ['Age', 'Salary']
imputer_mean = SimpleImputer(strategy='median')
df_median[numeric_columns] = imputer_mean.fit_transform(df_median[numeric_columns])

# Display the imputed DataFrames
print("\nImputed DataFrame using Median:")
print(df_median)
```

Remember, choose the right imputation technique based on your data and its characteristics.

Mean imputation and median imputation are two common methods for handling missing values in a dataset. The choice between them depends on the nature of the data and the specific characteristics of the variable with missing values.

#### Mean Imputation

- **When to Use**:
  - Use mean imputation when the data is normally distributed.
  - It's suitable for continuous data without significant outliers.
  - Appropriate when missing values are randomly distributed and represent a random sample of the data.

- **Advantages**:
  - Preserves the sample mean.
  - Simple and quick to implement.

- **Disadvantages**:
  - Can reduce the variability in the dataset.
  - Not suitable for skewed distributions.
  - Assumes that the mean is a good representation of the missing values, which might not always be the case.

#### Median Imputation

- **When to Use**:
  - Use median imputation for skewed data or when there are outliers.
  - Suitable for ordinal data (ranked data) or continuous data that is not normally distributed.
  - Effective when you want to preserve the central tendency of the data without being influenced by outliers.

- **Advantages**:
  - Less affected by outliers compared to mean imputation.
  - Preserves the median of the dataset.

- **Disadvantages**:
  - May not be as representative as mean for normally distributed data without outliers.
  - Like mean imputation, it can reduce the overall variability in the data.

##**Handling outliers: Detection and removal techniques**

### **Why outlier detection and removal is necessary?**

Outlier detection and removal is important because outliers can significantly affect your data analysis and the results of machine learning models. Think of outliers as very unusual points in your data, like a giraffe in a group of cats; they stand out because they're very different.

Here's why paying attention to these unusual points is important:

1. **Skewing the Results:** Outliers can skew your analysis. For example, if you're calculating the average height of a class of schoolchildren, and you accidentally include a professional basketball player, the average height will be much higher than it really should be.

2. **Misleading Information:** Outliers can give a false impression about trends or patterns in your data. They can make it seem like there's a relationship between variables when there isn't, or hide a real relationship that exists.

3. **Model Performance:** In machine learning, models try to learn patterns from the data. If outliers are present, the model might learn the wrong patterns. This is like trying to learn about cats but spending all your time looking at the giraffe. The model won't do a good job when it encounters typical data.

4. **Quality of Predictions:** Models influenced by outliers can make poor predictions. They're being misled by the unusual data, so their predictions might be too high, too low, or just irrelevant.

5. **Identifying Data Issues:** Sometimes outliers are caused by errors in data collection or processing (like a typo in entering data). Finding and removing them can improve the overall quality of your data.

6. **Understanding True Nature of Data:** By examining outliers, you can sometimes gain insights into the data. Maybe those outliers are telling you something important, like an unexpected behavior or a new trend.

However, it's important not to remove outliers just because they're outliers. Sometimes they contain valuable information about the dataset or about specific, unusual cases that are important to understand. The key is to investigate why they are outliers and then decide whether to keep them or remove them.

Sometimes, outliers just don't play well with our data analysis. So, it's time to evict those troublemakers! Let's see how we can effectively remove outliers! 💥👋

<u>**Example:**</u>

Let's consider the same Age and salary table:

```python
import pandas as pd
import numpy as np
from scipy import stats

# Create a DataFrame with some data including outliers
data = {
    'Age': [25, 32, 40, 28, 100, 35, 45, 52, 60],
    'Salary': [50000, 60000, 45000, 70000, 1000000, 55000, 48000, 52000, 80000]
}

df1 = pd.DataFrame(data)
df1
```

#### 3️⃣ **IQR Outlier Removal:** 📦❌

Use the Interquartile Range (IQR) to identify outliers, and say goodbye to those data points that fall outside the fences. It's like removing outliers using a sturdy data container!

```python
# IQR Outlier Removal
q1 = df1.quantile(0.25)
q3 = df1.quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers_iqr = df1[((df1 < lower_bound) | (df1 > upper_bound)).any(axis=1)]
df_no_outliers_iqr = df1[((df1 >= lower_bound) & (df1 <= upper_bound)).all(axis=1)]

print("\nOutliers detected using IQR Outlier Removal:")
print(outliers_iqr)
```

- The first quartile (q1), third quartile (q3), and interquartile range (iqr) are calculated using the **df.quantile()** function.
- The lower bound and upper bound are defined as  **(q1 - 1.5 * iqr)** and **(q3 + 1.5 * iqr)**, respectively.
- The DataFrame is filtered using **boolean indexing** to select rows where any value is below the lower bound or above the upper bound, indicating an outlier.
- The DataFrame without outliers is obtained by filtering rows where all values are within the specified range.

##**Dealing with categorical data: Techniques for encoding categorical variables**

Categorical data brings vibrant diversity to our analysis, but it requires special treatment. Let's explore some popular techniques for encoding categorical variables, using engaging descriptions! 🎨🔢



The "category encoder" package typically refers to a software library used in data science and machine learning for handling categorical data. Categorical data consists of variables that contain label values rather than numeric values. The primary function of a category encoder is to convert these label values into a form that can be more easily used in machine learning algorithms, which typically work better with numerical input.

The functions provided by a category encoder package can include:

1. **One-Hot Encoding:** Converts each category value into a new column and assigns a 1 or 0 (True/False) value to the column. This is useful for nominal data where no ordinal relationship exists.

2. **Ordinal Encoding:** Converts each unique category value into a numerical value. This is suitable for ordinal data where the categories have an inherent order.

3. **Binary Encoding:** This method combines the features of both one-hot and hash encoding. It first converts the categories into ordinal, then those ordinal numbers are converted into binary code, and then the digits from that binary string are split into separate columns.

4. **Hashing Encoding:** This method converts categories into a higher dimensional space of integers, where the distance between two categories in that space is arbitrary.

5. **Target Encoding:** This method replaces a categorical value with the mean of the target variable for that value.

These techniques are crucial for preparing categorical data for various machine learning models, as they can help in improving model performance by providing a meaningful way to represent category data for algorithms that require numerical input.

### **Why do we encode categorical variables?**

Encoding categorical data is essential in data science and machine learning because most algorithms work better with numerical data. Think of it like translating a book written in French into English, so more people can understand it.

Here's why encoding is important:

1. **Machine Learning Models Prefer Numbers:** Most machine learning models are like calculators; they understand and work best with numbers. So, we need to translate categories (like "red", "blue", "green") into numbers.

2. **Improves Model Accuracy:** Proper encoding can help the model understand the data better. This can lead to more accurate predictions, just like reading a book in a language you understand leads to better comprehension.

3. **Helps in Identifying Patterns:** When categorical data is converted into numbers, it becomes easier for models to detect patterns. It's like turning abstract concepts into concrete forms that are easier to analyze.

4. **Enables Mathematical Operations:** Numbers can be easily manipulated mathematically. This allows for various operations during model training, like calculating distances or averages, which isn't possible with text-based categories.

5. **Consistency in Data:** Encoding ensures that the same categories are represented consistently across the dataset, which is crucial for reliable analysis.

6. **Allows Comparison:** Numerical data can be compared and ranked, which isn’t always possible with categorical data. For instance, it’s easier to compare numbers than words when sorting or ordering.

In summary, encoding categorical data into numerical form makes it understandable and usable for machine learning models, much like translating a book into a widely spoken language makes it accessible to a larger audience.

##**Data normalization and scaling: Techniques and their applicability**

Data normalization and scaling are powerful techniques to bring our data into a consistent and manageable range. Let's explore different methods and their applicability. 🚀📊





### **Why feature scaling is required?**

Normalization or scaling of data is important, especially in machine learning and data analysis, because it's like making sure everyone uses the same measuring stick. Let's break this down into simpler terms:

Imagine you're comparing the height and weight of a group of people to understand their health. Heights might be measured in inches, ranging from, say, 60 to 72 inches, while weights could be in pounds, from 100 to 200 pounds. The problem is, these two measurements are on very different scales. If you try to compare them directly, the weight numbers are so much bigger that they'll dominate any analysis. It's like trying to compare the loudness of a whisper and a shout – the shout (or weight, in our example) drowns out the whisper (height).

Normalization or scaling adjusts all these different types of data to a common scale, without distorting the differences in the ranges of values. Here's why it's important:

1. **Equal Importance:** It ensures that each feature (like height and weight) is given equal importance. Without scaling, algorithms might incorrectly think that larger numbers are more important.

2. **Faster Learning:** Many machine learning algorithms converge faster (learn quicker) when data is on the same scale. It's like having a smoother road for the algorithm to travel on.

3. **Prevents Skewing:** Algorithms that compute distances between data points (like K-means clustering or K-nearest neighbors) can be skewed if one type of value ranges much higher or lower than others. Scaling prevents this.

4. **Improves Accuracy:** For many models, especially those involving neural networks or gradient descent optimization (like linear regression), scaling can lead to better accuracy in predictions.

5. **Required by Algorithms:** Some algorithms, like Support Vector Machines (SVM) and Principal Component Analysis (PCA), require scaled data to work properly.

In summary, normalization or scaling is about making sure that all the data you're using is speaking the same language, so to speak. It puts everything on an even playing field so that you can compare and use different types of data fairly and accurately in your analysis.

#### **Unit Vector Scaling:** 📐🔀



Unit vector scaling, or normalization, ensures that each data point has a magnitude of 1, preserving the relative proportions between variables. It's like making all our variables play on the same field and maintain their relationships!

```python
# Unit Vector Scaling
def unit_vector_scaling(x):
    norm = np.linalg.norm(x)
    if norm == 0:
        return x
    return x / norm

transformer_unit_vector = FunctionTransformer(unit_vector_scaling)
df_unit_vector = pd.DataFrame(transformer_unit_vector.transform(df1), columns=df1.columns)

# Print the original and scaled DataFrames
print("Original DataFrame:")
print(df1)
print("\nUnit Vector Scaling:")
print(df_unit_vector)
```

# **Try it yourself**

### **Task 1: Working on assignment**

**Problem Statemen**t: *Predicting California Housing Prices for Real Estate Market Analysis*

You work for a real estate company that aims to predict housing prices in California districts to gain insights into the housing market and assist customers in making informed decisions when buying or selling properties. Your task is to build a machine learning model that can accurately predict the median house value for California districts based on various features.

Dataset: "California Housing Prices" Dataset (Available in scikit-learn)

Project Tasks:

- Understanding the Data:

Explore the "California Housing Prices" dataset to understand its structure, data types, and distribution.
Identify the target variable, which is the "MedHouseValue" representing the median house value.
Analyze data quality, identify any data discrepancies, and handle any data cleaning required.

- Handling Missing Values:

Identify features with missing values and decide on an appropriate approach for handling them (e.g., imputation or removal).
Use imputation techniques like mean, median, or mode to fill in missing values in numerical features.
Use imputation techniques like most frequent value to fill in missing values in categorical features.

- Handling Outliers:

Detect outliers in numerical features that may affect the model's performance.
Decide on an appropriate approach for handling outliers, such as removing them or transforming them.

- Data Normalization and Scaling:

Examine the numerical features and identify whether normalization or scaling is required.
Apply normalization techniques like Min-Max scaling or Z-score scaling to bring numerical features to a similar scale.

- Dealing with Categorical Data (if applicable):

If categorical features exist in the dataset, apply techniques like one-hot encoding or label encoding to convert categorical variables into a numerical format.

- Building the Machine Learning Model:

Split the dataset into training and testing sets.
Select an appropriate regression algorithm (e.g., Linear Regression, Random Forest Regression, or XGBoost) to predict housing prices.
Train the model on the training data and evaluate its performance on the testing data.

- Model Evaluation and Fine-Tuning:

Use evaluation metrics like Mean Squared Error (MSE) or R-squared to assess the model's performance.
Fine-tune the model by adjusting hyperparameters and feature selection techniques to improve performance.

- Interpretation and Insights:

Analyze the model's coefficients or feature importances to understand the factors affecting housing prices.
Provide insights and recommendations for customers and stakeholders based on the model's predictions.
