# Exploratory Data Analysis (EDA)

**Definition**  
Exploratory Data Analysis (EDA) is a process in data science used to analyze datasets, summarize their key characteristics, and uncover patterns, anomalies, or relationships. It is essential for understanding the data before applying machine learning models or statistical techniques. EDA involves both numerical and graphical methods.

---

## Steps of EDA

### 1. **Understand Your Data**
- Review the dataset (rows, columns, and features).
- Identify variable types (numerical, categorical, or text).
- Check dataset dimensions (number of rows and columns).
- Inspect metadata (column names, data types, and descriptions).

### 2. **Clean the Data**
- Handle missing values (fill, drop, or impute).
- Fix incorrect data types (e.g., convert dates to `datetime`).
- Remove duplicates or irrelevant data.

### 3. **Explore Data with Summary Statistics**
- Calculate key statistics (mean, median, mode, standard deviation, variance).
- Analyze frequency distributions of categorical variables.

### 4. **Visualize the Data**
- **Univariate Analysis**: Histograms, box plots, and bar charts to examine single variables.
- **Bivariate Analysis**: Scatter plots, bar charts, and heatmaps for relationships between two variables.
- **Multivariate Analysis**: Pair plots, 3D plots, and heatmaps for relationships among multiple variables.

### 5. **Check for Outliers**
- Identify unusual values using box plots or Z-scores.
- Decide whether to handle outliers (e.g., remove, cap, or transform).

### 6. **Transform the Data**
- Normalize or standardize numerical data for consistent scaling.
- Encode categorical variables using one-hot encoding or label encoding.
- Engineer new features to enhance the dataset.

### 7. **Analyze Correlations**
- Use a correlation matrix (Pearson or Spearman) to study relationships between numerical variables.

### 8. **Document Insights**
- Summarize key patterns, trends, and anomalies.
- Highlight important findings and next steps for modeling or further analysis.

# **Check for Outliers**

Outliers are data points that are significantly different or far away from the rest of the data. They lie outside the expected range of values.

**Example:**

In a dataset of student ages [18, 19, 20, 21, 22, 50], the age 50 is an outlier.

**How to Detect Outliers**
1. Visualization:
* Box Plot: Shows outliers as points beyond whiskers.
* Scatter Plot: Reveals outliers in numerical relationships.
2. Statistical Methods:
* IQR Method:



**Handling Outliers**
* Remove: Exclude outliers if they are errors.
* Transform: Use log or square root transformations to reduce impact.
* Cap or Floor: Replace outliers with maximum or minimum acceptable values.
* Treat Separately: Analyze outliers independently for meaningful insights.


**IQR Method to Detect Outliers**
The Interquartile Range (IQR) Method is a statistical approach used to identify outliers in a dataset.
Steps to Use the IQR Method
Find the Quartiles:

Q1 (1st Quartile): The value below which 25% of the data lies.
Q3 (3rd Quartile): The value below which 75% of the data lies.
Calculate the IQR:

IQR=Q3−Q1

Determine the Outlier Boundaries:

Lower Bound:
Q1−1.5×IQR

Upper Bound:
Q3+1.5×IQR

**Identify Outliers:**

Any data point less than the lower bound or greater than the upper bound is considered an outlier.

# **IQR, Q1, and Q3 Explained**

---

## **1. Q1 (First Quartile)**
- **Definition**: The value below which **25% of the data** lies.
- **Purpose**: Marks the lower boundary of the middle 50% of the dataset.
- **Example**:
  - Dataset: [2, 4, 6, 8, 10]
  - Q1 = 4 (25% of values are less than or equal to 4).

---

## **2. Q3 (Third Quartile)**
- **Definition**: The value below which **75% of the data** lies.
- **Purpose**: Marks the upper boundary of the middle 50% of the dataset.
- **Example**:
  - Dataset: [2, 4, 6, 8, 10]
  - Q3 = 8 (75% of values are less than or equal to 8).

---

## **3. IQR (Interquartile Range)**
- **Definition**: The range of the middle 50% of the data, calculated as:

- **Purpose**: Measures the spread of the central portion of the data.
- **Example**:
- Dataset: [2, 4, 6, 8, 10]
- Q1 = 4, Q3 = 8, IQR = `8 - 4 = 4`.

---

# **Summary**
- **Q1**: Represents the lower quartile (25% of data).
- **Q3**: Represents the upper quartile (75% of data).
- **IQR**: Measures the spread of the central 50% of the data.

---

# **Use Case**
- **Outlier Detection**:
- Data points outside the range:
  ```
  [Q1 - 1.5 × IQR, Q3 + 1.5 × IQR]
  ```
- These values are considered **outliers**.


````python
# how to identify outliers
Q1 = df.loc[:,'price'].quantile(0.25)
Q3 = df.loc[:,'price'].quantile(0.75)
IQR = Q3-Q1
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
````

````python
lower_limit = Q1-1.5*IQR
upper_limit = Q3+1.5*IQR
print('lower_limit:',lower_limit)
print('Upper_limit:',upper_limit)
````

````python
df[df.loc[:,'price']<lower_limit]
````

````python
df[df.loc[:,'price']> upper_limit]
````

````python
df[(df.loc[:,'price']<lower_limit) | (df.loc[:,'price']> upper_limit)]
````

````python
# drop outliers since % is very less
# Identify rows where the 'price' is outside the specified range
outliers = df[(df['price'] < lower_limit) | (df['price'] > upper_limit)].index
# .index get indices of rows
# Drop the rows
df = df.drop(outliers)

# Reset the index after dropping rows (optional)
df.reset_index(drop=True, inplace=True)
# rest_index: Resets the DataFrame index to a continuous range, discarding the old index.
````

##**Shortcomings**

🤔 Biases and Confounding Variables: EDA can be impacted by biases and confounding variables that are present in the dataset. It's important to be aware of these factors and take them into consideration during the analysis.

👀 Limited Scope: EDA may be limited by the scope of the dataset, and may not be able to provide a comprehensive understanding of the problem or research question at hand.

🧪 Small Sample Size: EDA conducted on a small sample size may not be representative of the larger population, and can result in inaccurate or biased conclusions.

🤷‍♀️ Missing Data: EDA may be impacted by missing data that cannot be accurately imputed or estimated, which can lead to incomplete or inaccurate analysis.

📊 Limited Statistical Power: EDA conducted without appropriate statistical techniques or power may result in weak or inconclusive findings.

🕵️‍♀️ Limited Causality: EDA is exploratory in nature and cannot establish causality between variables, as it cannot control for all the variables that may impact the relationship between the variables of interest.

📉 Limited Predictive Power: EDA may not be able to provide accurate predictions or forecasts of future outcomes, as it focuses on exploring past data.

🚫 Data Privacy and Security: EDA can pose privacy and security risks if sensitive or personal information is present in the dataset. It's important to ensure that appropriate measures are taken to protect the data and maintain confidentiality.
